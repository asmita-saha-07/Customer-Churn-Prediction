```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pandas as pd 
import numpy as np
import pickle
```

```python
model=load_model('model.h5')
with open('label_encoder_gender.pkl','rb') as file:
    le=pickle.load(file)
with open('onehot_encodr_country.pkl','rb') as file:
    ohe=pickle.load(file)
with open('standard_scaler.pkl','rb') as file:
    scaler=pickle.load(file)
```

```python
#smple data
input={'credit_score':608,
'country':'Germany',
'gender':'Female',
'age':42,
'tenure':1,
'balance':83807.76,
'products_number':1,
'credit_card':0,
'active_member':1,
'estimated_salary':73931.40
}
```

```python
df=pd.DataFrame([input])
```

```python
df['gender']=le.transform(df['gender'])
cols=ohe.transform(df[['country']])
cols=pd.DataFrame(cols,columns=ohe.get_feature_names_out())
df=df.drop(['country'],axis=1)
df=pd.concat([df,cols],axis=1)
df=scaler.transform(df)
```

```python
pred=model.predict(df)
pred
```

```python
predict_proba=pred[0][0]
predict_proba
```

```python
if predict_proba>0.5:
    print("The person is likely to churn/leave")
else:
    print("The person is not likely to churn/leave")
```
